```r
library(MASS)     #simular normal bivariada
library(polycor)  #tetracorica/policorica
library(mvtnorm)
```

```r
set.seed(42)
rho_true <- 0.7 #Correlación real
n <- 500 #tamaño de la muestra
```

Variables latentes continuas (nunca las observamos en la practica)

$$\Sigma = \begin{pmatrix} 1 & \rho_{true} \\ \rho_{true} & 1 \end{pmatrix}$$


```r
Sigma <- matrix(c(1, rho_true, rho_true, 1), nrow = 2) #mu=0 varianza=1
latente <- mvrnorm(n, mu = c(0, 0), Sigma = Sigma) #v.l. continuas | corr 0.7
x_star <- latente[, 1]
y_star <- latente[, 2]
```

CASO 1: tetracorica (pues cortamos en 2 categorias)

```r
x_bin <- as.numeric(x_star > median(x_star)) #Continua -> Dicotómica con mediana
y_bin <- as.numeric(y_star > median(y_star))

r_pearson_bin <- cor(x_bin, y_bin) #Corr. Pearson
r_tetra <- polychor(x_bin, y_bin)  #Corr. Tetracórica

cat("Correlacion verdadera (latente):", rho_true, "\n")
cat("Pearson (2 categorias):         ", round(r_pearson_bin, 3), "\n")
cat("Tetracorica estimada:           ", round(r_tetra, 3), "\n")
```

CASO 2: policorica (pues cortamos en 5 categorias (quintiles))

```r
x_ord <- cut(x_star, breaks = quantile(x_star, probs = seq(0, 1, 0.2)), #1-5 Likert
             include.lowest = TRUE, labels = FALSE)
y_ord <- cut(y_star, breaks = quantile(y_star, probs = seq(0, 1, 0.2)),
             include.lowest = TRUE, labels = FALSE)

r_pearson_ord <- cor(x_ord, y_ord)
r_poly <- polychor(x_ord, y_ord)

cat("Pearson ingenuo (5 categorias):  ", round(r_pearson_ord, 3), "\n")
cat("Policorica estimada:             ", round(r_poly, 3), "\n")
```

phi vs Q vs tetracorica

```r
tabla <- matrix(c(40, 15, 10, 35), nrow = 2,
                 dimnames = list(X = c("0","1"), Y = c("0","1")))
a <- tabla[1,1]; b <- tabla[1,2]; c <- tabla[2,1]; d <- tabla[2,2]

phi <- (a*d - b*c) / sqrt((a+b)*(c+d)*(a+c)*(b+d))
Q   <- (a*d - b*c) / (a*d + b*c)

x2 <- rep(c(0,1), c(a+b, c+d))
y2 <- c(rep(c(0,1), c(a,b)), rep(c(0,1), c(c,d)))
r_tetra_ej <- polychor(x2, y2)

cat("phi:", round(phi,3), " | Q de Yule:", round(Q,3), " | tetracorica:", round(r_tetra_ej,3), "\n")
```
