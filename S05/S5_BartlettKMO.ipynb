```r
library(psych)
library(psychTools)
```

```r
#dataset msq
data(msq, package = "psychTools")
```

```r
cat("Dimensiones originales del dataset msq (filas, columnas):", dim(msq), "\n\n")
```

```r
colnames(msq)
```

```r
#Seleccionamos 12 aspectos emocionales para buscar factores latentes.
items_afe <- c("active", "alert", "aroused", "sleepy", "strong",
               "tense", "nervous", "calm", "happy", "sad", "tired", "placid")
```

```r
#Filtramos las columnas y eliminamos los valores faltantes (no puede tener NAN)
msq_subset <- msq[, items_afe]
msq_clean <- na.omit(msq_subset)

cat("Dimensiones después de la limpieza:", nrow(msq_clean), "filas, ",ncol(msq_clean), "columnas")
```

```r
#Cálculo de la matriz de correlación (Pearson)
matriz_cor <- cor(msq_clean)
```

```r
#Un p-valor < 0.05 rechaza la H0, demostrando que existe suficiente colinealidad
#entre los ítems para justificar la búsqueda de factores latentes.
cat("--- PRUEBA DE ESFERICIDAD DE BARTLETT ---\n")

bartlett_res <- cortest.bartlett(matriz_cor, n = nrow(msq_clean))

cat("El valor del estatistico es", bartlett_res$chisq, "con", bartlett_res$df, "grados de libertad y un p-value de",
    bartlett_res$p.value)
```

```r
#Índice de Adecuación Muestral KMO (Kaiser-Meyer-Olkin)
#¿Cuánta varianza en los ítems podría ser explicada por factores subyacentes?

cat("--- ÍNDICE KMO (Kaiser-Meyer-Olkin) ---\n")
kmo_res <- KMO(matriz_cor)
print(kmo_res)
```
