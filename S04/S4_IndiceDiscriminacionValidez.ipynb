```r
#Simulacion de un test dicotomico (TRO)

set.seed(2024)
n <- 200   #estudiantes
k <- 10    #items
theta <- rnorm(n)                                #habilidad real (no observada)
dificultad_b <- seq(-1.5, 1.5, length.out = k)   #dificultad de cada item, en escala z

respuestas <- sapply(dificultad_b, function(b) {
  ruido <- rnorm(n)
  as.numeric((theta + ruido) > b)                #1 = acierto, 0 = error
})
colnames(respuestas) <- paste0("Item", 1:k)
respuestas <- as.data.frame(respuestas)

puntaje_total <- rowSums(respuestas)

#Un criterio externo relacionado con theta, pero NO identico
#(ej: desempeño laboral, otra prueba, nota final del curso)
criterio_externo <- 0.6 * theta + rnorm(n, sd = sqrt(1 - 0.6^2))
```

```r
#Indice de discriminacion clasico (27% superior vs inferior)

grupo_superior <- puntaje_total >= quantile(puntaje_total, 0.73)
grupo_inferior <- puntaje_total <= quantile(puntaje_total, 0.27)

D <- sapply(respuestas, function(item) {
  p_superior <- mean(item[grupo_superior])
  p_inferior <- mean(item[grupo_inferior])
  p_superior - p_inferior
})
```

```r
#Indice de discriminacion: correlacion item-test corregida (punto-biserial)

r_item_total <- sapply(1:k, function(i) {
  total_sin_item <- puntaje_total - respuestas[, i]
  cor(respuestas[, i], total_sin_item)
})
names(r_item_total) <- colnames(respuestas)
```

```r
#Indice de validez (item vs criterio externo)

r_item_criterio <- sapply(1:k, function(i) {
  cor(respuestas[, i], criterio_externo)
})
names(r_item_criterio) <- colnames(respuestas)
```

```r
comparacion <- data.frame(
  Item = colnames(respuestas),
  Discriminacion_27 = round(D, 2),
  r_item_total_corregida = round(r_item_total, 2),
  Validez = round(r_item_criterio, 2)
)

comparacion
```
